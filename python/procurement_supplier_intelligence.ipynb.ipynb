```python
# Create a copy for cleaning
df_clean = df.copy()
```

```python
# Create Data Quality Report

quality_report = pd.DataFrame({

    "Column" : df_clean.columns,

    "Data Type" : df_clean.dtypes.values,

    "Missing Values" : df_clean.isnull().sum().values,

    "Missing %" : (
        df_clean.isnull().sum()/len(df_clean)*100
    ).round(2).values,

    "Unique Values" : df_clean.nunique().values

})

quality_report
```

```python
# Columns having missing values only

missing_table = quality_report[
    quality_report["Missing Values"] > 0
].sort_values(
    by="Missing %",
    ascending=False
)

missing_table
```

```python
# Plot missing values

plt.figure(figsize=(10,5))

plt.bar(
    missing_table["Column"],
    missing_table["Missing Values"]
)

plt.title("Missing Values Across Dataset")

plt.xlabel("Columns")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
categorical_cols = df_clean.select_dtypes(include="object").columns

for col in categorical_cols:
    df_clean[col] = (
        df_clean[col]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
date_cols = [col for col in df_clean.columns if "date" in col.lower()]

for col in date_cols:
    df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")

df_clean[date_cols].dtypes
```

```python
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print(df_clean.shape)
```

```python
numeric_cols = df_clean.select_dtypes(include=["int64","float64"]).columns

outlier_summary = []

for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    outliers = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()

    outlier_summary.append([col, outliers])

outlier_report = pd.DataFrame(outlier_summary, columns=["Column","Outliers"])

outlier_report.sort_values("Outliers", ascending=False)
```

```python
# Count null values
null_report = pd.DataFrame({
    "Column": df_clean.columns,
    "Null Count": df_clean.isnull().sum().values,
    "Null %": ((df_clean.isnull().sum()/len(df_clean))*100).round(2).values
})

null_report = null_report[null_report["Null Count"] > 0]
null_report.sort_values("Null %", ascending=False)
```

```python
# Identify column types
numeric_cols = df_clean.select_dtypes(include=["int64","float64"]).columns
categorical_cols = df_clean.select_dtypes(include="object").columns
date_cols = df_clean.select_dtypes(include="datetime64[ns]").columns

# Fill numeric columns with Median
for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Fill categorical columns with Mode
for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

# Fill date columns using Forward Fill
for col in date_cols:
    df_clean[col] = df_clean[col].fillna(method="ffill")
```

```python
# Final null check
df_clean.isnull().sum()
```

```python
status_count = df_clean["Order_Status"].value_counts()

plt.figure(figsize=(8,5))
plt.bar(status_count.index, status_count.values)

plt.title("Purchase Orders by Order Status")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
supplier_count = df_clean["Supplier"].value_counts()

plt.figure(figsize=(10,5))
plt.bar(supplier_count.index, supplier_count.values)

plt.title("Purchase Orders by Supplier")
plt.xlabel("Supplier")
plt.ylabel("Number of Purchase Orders")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_quantity = (
    df_clean.groupby("Item_Category")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))
plt.bar(category_quantity.index, category_quantity.values)

plt.title("Procurement Quantity by Item Category")
plt.xlabel("Item Category")
plt.ylabel("Total Quantity")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
supplier_quantity = (
    df_clean.groupby("Supplier")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))
plt.bar(supplier_quantity.index, supplier_quantity.values)

plt.title("Total Procurement Quantity by Supplier")
plt.xlabel("Supplier")
plt.ylabel("Total Quantity")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df_clean["Unit_Price"], bins=30)

plt.title("Distribution of Unit Prices")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df_clean["Negotiated_Price"], bins=30)

plt.title("Distribution of Negotiated Prices")
plt.xlabel("Negotiated Price")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
supplier_defects = (
    df_clean.groupby("Supplier")["Defective_Units"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

plt.bar(
    supplier_defects.index,
    supplier_defects.values
)

plt.title("Defective Units by Supplier")
plt.xlabel("Supplier")
plt.ylabel("Defective Units")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_clean["Order_Date"] = pd.to_datetime(df_clean["Order_Date"])
df_clean["Delivery_Date"] = pd.to_datetime(df_clean["Delivery_Date"])

df_clean["Delivery_Days"] = (
    df_clean["Delivery_Date"] -
    df_clean["Order_Date"]
).dt.days

plt.figure(figsize=(8,5))

plt.hist(df_clean["Delivery_Days"].dropna(), bins=30)

plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Days")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```

```python
supplier_performance = (
    df_clean.groupby("Supplier")
    .agg(
        Avg_Delivery_Days=("Delivery_Days", "mean"),
        Total_Quantity=("Quantity", "sum"),
        Total_Defects=("Defective_Units", "sum")
    )
)

supplier_performance["Defect_Rate"] = (
    supplier_performance["Total_Defects"] /
    supplier_performance["Total_Quantity"] * 100
)

plt.figure(figsize=(9,6))

plt.scatter(
    supplier_performance["Avg_Delivery_Days"],
    supplier_performance["Defect_Rate"]
)

plt.title("Supplier Delivery Time vs Defect Rate")
plt.xlabel("Average Delivery Days")
plt.ylabel("Defect Rate (%)")

plt.tight_layout()
plt.show()
```
